# Travel Blog & Gallery: Ocean Theme
This notebook sets up a Flask web application for a travel blog featuring Dominican Republic, Rome, Milan, London, Barcelona, and Chicago.

```python
import os

# Create directory structure
os.makedirs('templates', exist_ok=True)
os.makedirs('static/css', exist_ok=True)

# Define the CSS with an ocean/aqua blue theme
css_content = """
body {
    font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
    background-color: #e0f7fa;
    color: #006064;
    margin: 0;
    padding: 0;
}
header {
    background-color: #00bcd4;
    color: white;
    padding: 2rem;
    text-align: center;
}
nav {
    background-color: #0097a7;
    padding: 1rem;
    text-align: center;
}
nav a {
    color: white;
    margin: 0 15px;
    text-decoration: none;
    font-weight: bold;
}
.container {
    padding: 20px;
    max-width: 1000px;
    margin: auto;
}
.post-card {
    background: white;
    border-radius: 8px;
    padding: 20px;
    margin-bottom: 20px;
    box-shadow: 0 4px 8px rgba(0,0,0,0.1);
}
.gallery {
    display: grid;
    grid-template-columns: repeat(auto-fit, minmax(250px, 1fr));
    gap: 15px;
}
.photo-placeholder {
    background-color: #b2ebf2;
    height: 200px;
    display: flex;
    align-items: center;
    justify-content: center;
    border-radius: 4px;
}
"""

with open('static/css/style.css', 'w') as f:
    f.write(css_content)

print('CSS and directories created.')
```

```python
locations = ['Dominican Republic', 'Rome', 'Milan', 'London', 'Barcelona', 'Chicago']

base_html = """
<!DOCTYPE html>
<html>
<head>
    <link rel="stylesheet" href="{{ url_for('static', filename='css/style.css') }}">
    <title>Ocean Travel Blog</title>
</head>
<body>
    <header><h1>Aqua Travels</h1></header>
    <nav>
        <a href="/">Home</a>
        {% for loc in locations %}
        <a href="/post/{{ loc.lower().replace(' ', '_') }}">{{ loc }}</a>
        {% endfor %}
    </nav>
    <div class="container">
        {% block content %}{% endblock %}
    </div>
</body>
</html>
"""

with open('templates/base.html', 'w') as f:
    f.write(base_html)

# Home page template
home_html = """
{% extends 'base.html' %}
{% block content %}
<h2>Welcome to the Horizon</h2>
<p>Explore our latest adventures across the globe.</p>
<div class="gallery">
    {% for loc in locations %}
    <div class="post-card">
        <h3>{{ loc }}</h3>
        <a href="/post/{{ loc.lower().replace(' ', '_') }}">View Gallery</a>
    </div>
    {% endfor %}
</div>
{% endblock %}
"""

with open('templates/index.html', 'w') as f:
    f.write(home_html)

# Post/Gallery template
post_html = """
{% extends 'base.html' %}
{% block content %}
<h2>{{ location }} Gallery</h2>
<p>Capturing the essence of {{ location }}.</p>
<div class="gallery">
    <div class="photo-placeholder">Photo 1</div>
    <div class="photo-placeholder">Photo 2</div>
    <div class="photo-placeholder">Photo 3</div>
</div>
{% endblock %}
"""

with open('templates/post.html', 'w') as f:
    f.write(post_html)

print('Templates created.')
```

```python
from flask import Flask, render_template

app = Flask(__name__)

LOCATIONS = ['Dominican Republic', 'Rome', 'Milan', 'London', 'Barcelona', 'Chicago']

@app.route('/')
def index():
    return render_template('index.html', locations=LOCATIONS)

@app.route('/post/<name>')
def post(name):
    display_name = name.replace('_', ' ').title()
    return render_template('post.html', location=display_name, locations=LOCATIONS)

if __name__ == '__main__':
    print("Flask app defined. To run this in Colab, you would typically use a library like flask-ngrok or serve it via a local tunnel.")
```

```python
import threading

def run_app():
    # Run the flask app on 0.0.0.0 and a new port
    app.run(host='0.0.0.0', port=5002, debug=False, use_reloader=False)

# Start Flask in a background thread
threading.Thread(target=run_app).start()
print('Flask server is running on http://0.0.0.0:5002')
```
